```python
SYSTEM_PROMPT = """
You are CampusBot, an AI assistant for ABC University.

Your job is to help students with:
- Course registration
- Exam schedules
- Fee payments
- Attendance requirements
- Library services

Tone:
Friendly
Professional
Simple English
"""
```

```python
def ask(user_message):

    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role":"system",
                "content":SYSTEM_PROMPT
            },
            {
                "role":"user",
                "content":user_message
            }
        ]
    )

    return response["message"]["content"]
```

```python
conversation_history = [
    {
        "role":"system",
        "content":SYSTEM_PROMPT
    }
]
```

```python
def chat(user_message):

    conversation_history.append(
        {
            "role":"user",
            "content":user_message
        }
    )

    response = ollama.chat(
        model="llama3.2:3b",
        messages=conversation_history
    )

    reply = response["message"]["content"]

    conversation_history.append(
        {
            "role":"assistant",
            "content":reply
        }
    )

    return reply
```

```python
SYSTEM_PROMPT = """
You are CampusBot, an AI assistant for ABC University.

Your job is to help students with:
- Course registration
- Exam schedules
- Fee payments
- Attendance requirements
- Library services

Boundaries:
- Do NOT provide legal advice.
- Do NOT provide tax advice.
- Do NOT discuss competitor universities.
- If the request is outside your scope, politely ask the student to contact the university office.

Tone:
Friendly
Professional
Simple English
"""
```

```python
BLOCKED_TOPICS = [
    "lawsuit",
    "tax advice",
    "competitor university"
]
```

```python
def guarded_chat(user_message):

    if any(
        topic in user_message.lower()
        for topic in BLOCKED_TOPICS
    ):

        return "This question is outside my scope. Please contact the university administration."

    return chat(user_message)
```

```python
print(guarded_chat("Should I file a lawsuit against my landlord?"))
```

```python
print(guarded_chat("Which university is better than ABC University?"))
```

```python
def escalate(user_message, reason):

    print(f"[ESCALATED] Reason: {reason}")

    return "I am forwarding your request to a university staff member."
```

```python
def assistant_turn(user_message):

    if any(
        topic in user_message.lower()
        for topic in BLOCKED_TOPICS
    ):

        return escalate(
            user_message,
            "Out of Scope"
        )

    reply = chat(user_message)

    uncertainty = [
        "i'm not sure",
        "i don't know",
        "i can't verify"
    ]

    if any(
        text in reply.lower()
        for text in uncertainty
    ):

        return escalate(
            user_message,
            "Low Confidence"
        )

    return reply
```

```python
print(assistant_turn("Why was my fee payment rejected yesterday?"))
```

```python
print(assistant_turn("Why are my exam marks incorrect?"))
```

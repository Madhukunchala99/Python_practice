```python
my_str = "India is my country and i love india"

count = 0

for ch in my_str.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels:", count)
```

```python
for num in range(1, 1001):
    temp = num
    rev = 0

    while temp > 0:
        digit = temp % 10
        rev = rev * 10 + digit
        temp = temp // 10

    if num == rev:
        print(num)
```

```python
n = int(input('Enter a number: '))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n//10
print(sum)    
```

```python
my_str = "India is my country"

word = ""
result = []

for ch in my_str:
    if ch != " ":
        word = word + ch
    else:
        result.append(word)
        word = ""

result.append(word)

print(result)
```

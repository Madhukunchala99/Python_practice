```python
# ==========================================
# Function: np.add()
#
# Usage:
# Adds two arrays or numbers element-wise.
#
# Syntax:
# np.add(x1, x2)
#
# Example:
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

result = np.add(a, b)

print("Array 1:", a)
print("Array 2:", b)
print("After Addition:", result)
```

```python
# ==========================================
# Function: np.all()
#
# Usage:
# Checks whether all elements in an array satisfy a condition.
# Returns True if all elements are True (or non-zero), otherwise False.
#
# Syntax:
# np.all(a)
#
# Example:
import numpy as np

arr = np.array([10, 20, 30, 40])

result = np.all(arr > 5)

print("Array:", arr)
print("Are all elements greater than 5?")
print(result)
```

```python
# ==========================================
# Function: np.allclose()
#
# Usage:
# Checks whether two arrays are element-wise equal within a specified tolerance.
# Returns True if all corresponding elements are close enough; otherwise False.
#
# Syntax:
# np.allclose(a, b, rtol=1e-05, atol=1e-08)
#
# Example:
import numpy as np

arr1 = np.array([1.0, 2.0, 3.0])
arr2 = np.array([1.000001, 2.000001, 3.000001])

result = np.allclose(arr1, arr2)

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Are the arrays approximately equal?")
print(result)
```

```python
# ==========================================
# Function: np.any()
#
# Usage:
# Checks whether at least one element in an array satisfies a condition.
# Returns True if any element is True (or non-zero); otherwise False.
#
# Syntax:
# np.any(a)
#
# Example:
import numpy as np

arr = np.array([10, -5, 20, 30])

result = np.any(arr < 0)

print("Array:", arr)
print("Is there any negative number in the array?")
print(result)
```

```python
# ==========================================
# Function: np.tri()
# Usage:
# Creates an array with ones at and below the main diagonal and zeros above it.
# It is used for creating lower triangular matrices in NumPy.
# Syntax:
# np.tri(...)

import numpy as np

print("Function:", "np.tri")

# Simple example
result = np.tri(3, 3)

print(result)
```

```python
# ==========================================
# Function: np.unique()
# Usage:
# Finds the unique elements from a NumPy array by removing duplicate values.
# It is used to identify distinct values and simplify data analysis.
# Syntax:
# np.unique(...)

import numpy as np

print("Function:", "np.unique")

# Simple example
arr = np.array([1, 2, 2, 3, 4, 4, 5])

result = np.unique(arr)

print(result)
```

```python
# ==========================================
# Function: np.where()
# Usage:
# Finds the positions of elements that satisfy a given condition.
# It is used for filtering and replacing values in NumPy arrays.
# Syntax:
# np.where(...)

import numpy as np

print("Function:", "np.where")

# Simple example
arr = np.array([10, 25, 5, 30, 15])

result = np.where(arr > 20)

print(result)
```

```python
# ==========================================
# Function: np.zeros()
# Usage:
# Creates a NumPy array filled with zeros.
# It is used to initialize arrays with zero values before storing data.
# Syntax:
# np.zeros(...)

import numpy as np

print("Function:", "np.zeros")

# Simple example
arr = np.zeros((2, 2))

print(arr)
```

```python
# ==========================================
# Function: np.zeros_like()
# Usage:
# Creates an array of zeros with the same shape and data type as an existing array.
# It is used when we need a zero-filled array similar to another array.
# Syntax:
# np.zeros_like(...)

import numpy as np

print("Function:", "np.zeros_like")

# Simple example
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

result = np.zeros_like(arr)

print(result)
```

```python
# ==========================================
# Function: np.__version__
# Usage:
# Displays the installed version of the NumPy library.
# It is used to check which NumPy version is currently available in the system.
# Syntax:
# np.__version__

import numpy as np

print("Function:", "np.__version__")

# Simple example
print(np.__version__)
```
